```python
import pandas as pd
from sklearn.model_selection import train_test_split
import torch
from torch.utils.data import Dataset, DataLoader
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt
```

```python
torch.manual_seed(42)
```

```python
df=pd.read_csv("/content/fmnist_small (1).csv")
df
```

```python
X=df.iloc[:,1:].values
y=df.iloc[:,0].values
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.20,random_state=42)
```

```python
X_train=X_train/255.0
X_test=X_test/255.0
```

```python
class CustomData(Dataset):

  def __init__(self,X,y):
    self.X= torch.tensor(X, dtype=torch.float32)
    self.y= torch.tensor(y, dtype=torch.long)

  def __len__(self):

    return len(self.X)

  def __getitem__(self,indx):

    return self.X[indx],self.y[indx]
```

```python
train_dataset=CustomData(X_train,y_train)
test_dataset=CustomData(X_test,y_test)
```

```python
train_loader=DataLoader(train_dataset,batch_size=32,shuffle=True)
```

```python
test_loader=DataLoader(test_dataset,batch_size=32,shuffle=False)
```

```python
class MyNN(nn.Module):

  def __init__(self,input_dim,output_dim,nuerons_per_layer,num_hidden_layer,dropout_prob):

    super().__init__()
    layer=[]

    for i in range(num_hidden_layer):

      layer.append(nn.Linear(input_dim,nuerons_per_layer))
      layer.append(nn.ReLU())
      layer.append(nn.Dropout(p=dropout_prob))
      input_dim=nuerons_per_layer

    layer.append(nn.Linear(nuerons_per_layer,output_dim))
    self.model=nn.Sequential(*layer)

  def forward(self,feature):

    return self.model(feature)
```

```python
def objective(trial):

  nuerons_per_layer=trial.suggest_int("nuerons_per_layer",8,128,step=8)
  num_hidden_layer=trial.suggest_int("num_hidden_layer",1,5)
  dropout_prob=trial.suggest_float("dropout_prob",0.1,0.6,step=0.1)
  epochs=trial.suggest_int("epochs",10,50,step=10)
  learning_rate=trial.suggest_float("learning_rate",1e-5,1e-1,log=True)
  weight_decay=trial.suggest_float("weight_decay",1e-5,1e-3,log=True)
  optimizer_name=trial.suggest_categorical("optimizer_name",["SGD","Adam","RMSprop"])
  batch_size=trial.suggest_categorical("batch_size",[16,32,64,128])

  input_dim=784
  output_dim=10

  train_loader=DataLoader(train_dataset,batch_size=batch_size,shuffle=True)
  test_loader=DataLoader(test_dataset,batch_size=batch_size,shuffle=False)

  model=MyNN(input_dim,output_dim,nuerons_per_layer,num_hidden_layer,dropout_prob)

  #params init
  criteria=nn.CrossEntropyLoss()
  optimizer = optim.SGD(model.parameters(), lr=0.1, weight_decay=1e-4)

  #optimizer selection

  if optimizer_name=="SGD":
    optim.SGD(model.parameters(),learning_rate,weight_decay)
  elif optimizer_name=="Adam":
    optim.Adam(model.parameters(),learning_rate,weight_decay)
  else:
    optim.RMSprop(model.parameters(),learning_rate,weight_decay)

  #training loop
  for i in range(epochs):

    for batch_features,batch_labels in train_loader:

      output=model(batch_features)

      loss=criteria(output,batch_labels)

      optimizer.zero_grad()
      loss.backward()

      optimizer.step()

  #evaluation
  model.eval()
  correct=0
  total=0
  with torch.no_grad():

    for batch_features,batch_labels in test_loader:

        output=model(batch_features)

        _,prediction=torch.max(output,1)

        correct=correct+(batch_labels==prediction).sum().item()

        total+=output.shape[0]
    accuracy=correct/total

  return accuracy
```

```python
study.optimize(objective,n_trials=10)
```

```python
class MyNN(nn.Module):

  def __init__(self,features):
    super().__init__()

    self.model=nn.Sequential(
        nn.Linear(features,128),
        nn.BatchNorm1d(128),
        nn.ReLU(),
        nn.Dropout(p=0.3),
        nn.Linear(128,64),
        nn.BatchNorm1d(64),
        nn.ReLU(),
        nn.Dropout(p=0.3),
        nn.Linear(64,10)
    )

  def forward(self,x):

    return self.model(x)
```

```python
model=MyNN(X_train.shape[1])
```

```python
criteria=nn.CrossEntropyLoss()
optimizer=optim.SGD(model.parameters(),lr=.1,weight_decay=1e-4)
```

```python
for epoch in range(100):
  total_loss=0
  for batch_features,batch_labels in train_loader:

    output=model(batch_features)

    loss=criteria(output,batch_labels)

    optimizer.zero_grad()
    loss.backward()

    optimizer.step()

    total_loss+=loss

  print(f"epoch{epoch+1} :{total_loss}")
```
